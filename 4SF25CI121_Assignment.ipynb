## 3. Question
Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.

```python
df3 = pd.DataFrame(np.random.randint(1, 20, size=(5, 3)), columns=['A', 'B', 'C']).astype(float)
df3.iloc[0, 1] = np.nan
df3.iloc[2, 0] = np.nan
df3.iloc[4, 2] = np.nan
print("With NaN:\n", df3, "\n")
df3_filled = df3.fillna(df3.mean())
print("Filled with column means:\n", df3_filled)
```

## 5. Data Aggregation
Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.

```python
df5 = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], size=10),
    'Value': np.random.randint(1, 100, size=10)
})
print(df5, "\n")
df5.groupby('Category')['Value'].agg(['sum', 'mean'])
```

## 6. Question
Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.

```python
df6 = pd.DataFrame({
    'Product': [f'P{i}' for i in range(1, 9)],
    'Category': np.random.choice(['Electronics', 'Clothing', 'Food'], size=8),
    'Sales': np.random.randint(100, 1000, size=8)
})
print(df6, "\n")
df6.groupby('Category')['Sales'].sum()
```

## 7. Merging DataFrames
Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

```python
left = pd.DataFrame({'ID': [1, 2, 3, 4], 'Name': ['Asha', 'Ravi', 'Meena', 'Karan']})
right = pd.DataFrame({'ID': [2, 3, 4, 5], 'Score': [88, 92, 75, 60]})

print("Inner merge:")
display(pd.merge(left, right, on='ID'))
print("Outer merge:")
display(pd.merge(left, right, on='ID', how='outer'))
```

## 8. MultiIndex DataFrame
Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame.

```python
index8 = pd.MultiIndex.from_product([['X', 'Y'], [1, 2, 3]], names=['Letter', 'Number'])
df8 = pd.DataFrame(np.random.randint(1, 50, size=(6, 2)), index=index8, columns=['Col1', 'Col2'])
df8
```

```python
# All rows for outer label 'X'
df8.loc['X']
```

```python
# Single row by full key
df8.loc[('Y', 2)]
```

```python
# Cross-section: Number == 1 across all letters
df8.xs(1, level='Number')
```

```python
# Slice from ('X', 1) to ('Y', 2) on Col1
df8.loc[('X', 1):('Y', 2), 'Col1']
```

## 9. Question
Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.

```python
index9 = pd.MultiIndex.from_tuples(
    [('Fruit', 'Apple'), ('Fruit', 'Banana'), ('Fruit', 'Mango'),
     ('Veg', 'Carrot'), ('Veg', 'Potato'), ('Veg', 'Onion')],
    names=['Category', 'SubCategory'])
df9 = pd.DataFrame({'Value': np.random.randint(10, 100, size=6)}, index=index9)
df9
```

```python
# Sum per Category
df9.groupby(level='Category').sum()
```

```python
# Sum per Category and SubCategory
df9.groupby(level=['Category', 'SubCategory']).sum()
```

## 10. Applying Functions
Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Apply a function that doubles the values of the DataFrame.

```python
df10 = pd.DataFrame(np.random.randint(1, 20, size=(5, 3)), columns=['A', 'B', 'C'])

def double(x):
    return x * 2

print("Original:\n", df10, "\n")
print("Doubled:\n", df10.apply(double))
# Vectorized equivalent (fastest): df10 * 2
```

## 11. Question
Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers. Apply a lambda function to create a new column that is the sum of the existing columns.

```python
df11 = pd.DataFrame(np.random.randint(1, 20, size=(6, 3)), columns=['A', 'B', 'C'])
df11['Total'] = df11.apply(lambda row: row['A'] + row['B'] + row['C'], axis=1)
df11
# Vectorized equivalent: df11[['A','B','C']].sum(axis=1)
```
